```python
import os
import cv2
import numpy as np
import pandas as pd

def process_images(image_dir, label, data_list):
    for img_file in os.listdir(image_dir):
        img_path = os.path.join(image_dir, img_file)
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Read image in grayscale
        if img is None:
            continue

        img_resized = cv2.resize(img, (48, 48))  # Resize to 48x48 pixels
        pixels = img_resized.flatten()  # Flatten to 1D array of 2304 pixels (48x48)
        pixels_str = ' '.join(map(str, pixels))  # Convert to space-separated string

        data_list.append([label, pixels_str])

# Function to create CSV file for FER dataset
def create_fer_csv(fer_dir, output_csv):
    data = []
    for folder in ['train', 'test']:
        folder_path = os.path.join(fer_dir, folder)
        for label_dir in os.listdir(folder_path):
            label = label_dir  # Assuming folder name is the label
            img_dir = os.path.join(folder_path, label_dir)
            process_images(img_dir, label, data)


    df = pd.DataFrame(data, columns=['label', 'pixels'])
    df.to_csv(output_csv, index=False)
    print(f"FER CSV saved: {output_csv}")

ck_class_mapping = {
    'anger': 0,
    'disgust': 1,
    'fear': 2,
    'happy': 3,
    'sad': 4,
    'surprise': 5,
    'neutral': 6
}

def create_ck_csv(ck_dir, output_csv):
    data = []
    for class_dir in os.listdir(ck_dir):
        if class_dir.lower() not in ck_class_mapping:
            continue

        label = ck_class_mapping[class_dir.lower()]
        img_dir = os.path.join(ck_dir, class_dir)
        process_images(img_dir, label, data)

    # Save to CSV
    df = pd.DataFrame(data, columns=['label', 'pixels'])
    df.to_csv(output_csv, index=False)
    print(f"CK CSV saved: {output_csv}")

# Now call the function to create the CSV
create_ck_csv(ck_dataset_dir, ck_csv_output)


# Paths to datasets
fer_dataset_dir = '/content/fer'
ck_dataset_dir = '/content/ck'

# Output CSV files
fer_csv_output = '/content/fer_data.csv'
ck_csv_output = '/content/ck_data.csv'

# Create CSV files
create_fer_csv(fer_dataset_dir, fer_csv_output)
create_ck_csv(ck_dataset_dir, ck_csv_output)
```
